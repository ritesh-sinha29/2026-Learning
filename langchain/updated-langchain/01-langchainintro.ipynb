```python
## Define the Custom Tool
from langchain_core.tools import tool

@tool
def multiply_numbers(a: int, b: int) -> str:
    """Multiplies two integers together. Use this tool when you need to calculate the product of two numbers.
    
    Args:
        a: The first integer
        b: The second integer
    """
    result = a * b
    return f"The result of {a} multiplied by {b} is {result}."
```

```python
from langchain.agents import create_agent
from langchain_tavily import TavilySearch

# Initialize both tools
web_search_tool = TavilySearch()
custom_multiply_tool = multiply_numbers

# Create the agent with both tools
agent = create_agent(
    model=llm,
    tools=[web_search_tool, custom_multiply_tool],
    system_prompt="You are a helpful assistant. Use tools when needed.",
    debug=True
)
```

```python
response = agent.invoke({"messages": [("user", "What is 457 multiplied by 82?")]})

# Print the agent's final answer
print(response["messages"][-1].content)
```

```python
# We import 'tool' from langchain_core. This decorator wraps a standard Python
# function and turns it into a LangChain 'BaseTool' object.
from langchain_core.tools import tool

# 1. The @tool decorator registers this function.
@tool
def calculate_factorial(n: int) -> str:
    # 2. DOCSTRING (CRITICAL): The LLM reads this description to decide 
    #    WHEN it should call this tool. If you omit the docstring, 
    #    the agent will not know what the tool is used for.
    """Calculates the factorial of a given integer.
    
    Use this tool when the user asks for the factorial of a number.
    
    Args:
        n: The integer to compute the factorial for. Must be a non-negative number.
    """
    
    # 3. TYPE HINTS (n: int): LangChain reads these type hints to generate
    #    the JSON input schema. The LLM will use this schema to construct
    #    the arguments. In this case, it ensures the LLM passes an integer.
    
    # 4. Internal logic (What the tool actually does):
    if n < 0:
        return "Error: Factorial is not defined for negative numbers."
        
    import math
    result = math.factorial(n)
    
    # 5. The return value MUST be a string (or serialize to one) so the agent
    #    can read it and pass it back to the LLM as text context.
    return f"The factorial of {n} is {result}."
```

```python
# Pydantic is used to define structured input schemas.
from pydantic import BaseModel, Field
from typing import Literal
from langchain_core.tools import tool

# 1. Define the Schema Class:
#    This class specifies exactly what inputs the tool accepts.
class WeatherInput(BaseModel):
    # Field(...) lets you add description metadata to each variable.
    # The LLM reads these descriptions to understand what value to pass.
    location: str = Field(
        description="The city name to check weather for, e.g. 'Paris', 'Tokyo'."
    )
    
    # Literal["celsius", "fahrenheit"] constrains the LLM's options.
    # The model is forced to choose ONLY one of these two strings.
    units: Literal["celsius", "fahrenheit"] = Field(
        default="celsius",
        description="The temperature unit preference."
    )
    
    include_forecast: bool = Field(
        default=False,
        description="Whether to return a 3-day forecast instead of just today's weather."
    )

# 2. Apply the decorator, linking the schema via 'args_schema':
#    This tells LangChain: "Validate inputs against the WeatherInput model before running."
@tool(args_schema=WeatherInput)
def get_weather(location: str, units: str = "celsius", include_forecast: bool = False) -> str:
    """Gets the current weather and forecast for a specific location."""
    
    # Mock data lookup (internal execution code)
    temp = 21 if units == "celsius" else 70
    condition = "Partly Cloudy"
    
    result = f"Current weather in {location}: {temp}°{units[0].upper()} and {condition}."
    
    if include_forecast:
        result += f"\nForecast: Day 1: Sunny | Day 2: Overcast | Day 3: Rain"
        
    return result
```

```python
import json
from langchain_core.tools import tool

@tool
def count_vowels_and_consonants(text: str) -> str:
    """Counts the number of vowels, consonants, and spaces in a given string.
    
    Args:
        text: The input text to analyze.
    """
    vowels = "aeiouAEIOU"
    v_count = sum(1 for char in text if char in vowels)
    c_count = sum(1 for char in text if char.isalpha() and char not in vowels)
    space_count = sum(1 for char in text if char.isspace())
    
    analysis = {
        "vowels": v_count,
        "consonants": c_count,
        "spaces": space_count,
        "total_characters": len(text)
    }
    return json.dumps(analysis, indent=2)
```

```python
from langchain_core.tools import tool

# Mock database
USER_ACCOUNTS = {
    "alice": {"balance": 1500.50, "status": "active"},
    "bob": {"balance": 45.00, "status": "suspended"},
    "charlie": {"balance": 350.00, "status": "active"}
}

@tool
def check_user_balance(username: str) -> str:
    """Retrieves the current account balance and status for a given user.
    
    Args:
        username: The lowercase name of the user (e.g., 'alice').
    """
    user = username.strip().lower()
    if user in USER_ACCOUNTS:
        info = USER_ACCOUNTS[user]
        return f"User {username} has a balance of ${info['balance']:.2f} and account is {info['status']}."
    return f"User '{username}' was not found in the database."
```

```python
# Pass multiple custom tools to the agent
agent = create_agent(
    model=llm,
    tools=[check_user_balance, get_weather, calculate_factorial, count_vowels_and_consonants],
    system_prompt="You are a helpful assistant with access to weather, account, and text analysis tools.",
    debug=True
)
```

```python
agent.invoke({"messages":"What is 457 multiplied by 82?"})
```
